```python
df['Datetime'] = pd.to_datetime(df['Datetime'])
print(df['Datetime'].dt.year.value_counts())
```

Avoid Data Leakage
1. Train set (2004-2011) is clean historical training
2. Eval set (2012-2015) is unseen during training, but used for validation and hyperparameter tuning
3. Holdout set (2016-2018) is completely untouched until the very end

```python
# Sort by date (important to avoid shuffling issues)
df = df.sort_values("Datetime")
```

```python
# Define cutoff dated
cutoff_date_eval = "2012-01-01" # validation start
cutoff_date_holdout = "2016-01-01"
```

```python
# Train: before 2012
train_df = df[df["Datetime"] < cutoff_date_eval]

# Validation/Eval: 2012 - 2015
eval_df = df[
    (df["Datetime"] >= cutoff_date_eval) & 
    (df["Datetime"] < cutoff_date_holdout)
]

# Holdout: 2016 - 2018
holdout_df = df[df["Datetime"] >= cutoff_date_holdout]

print("Train shape:", train_df.shape)
print("Eval shape:", eval_df.shape)
print("Holdout shape", holdout_df.shape)
```

```python
# Save splits
train_df.to_csv(r"C:\Users\sheri\Desktop\machinlearning\data\raw_data\train.csv", index=False)
eval_df.to_csv(r"C:\Users\sheri\Desktop\machinlearning\data\raw_data\eval.csv", index=False)
holdout_df.to_csv(r"C:\Users\sheri\Desktop\machinlearning\data\raw_data\holdout.csv", index=False)
```
